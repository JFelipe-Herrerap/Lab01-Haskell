```haskell
productoLista :: [Int] -> Int
productoLista [] = 1
productoLista(x:xs) = x * productoLista xs
```

```haskell
productoLista [1,2,3,4,5]
```

```haskell
eliminarNegativos :: [Int] -> [Int]
eliminarNegativos [] = [] 
eliminarNegativos (x:xs) 
   |x >= 0 = x : eliminarNegativos xs
   
```

```haskell
eliminarNegativos [-10,3,5,-8,-11,7]
```

```haskell
invertir :: [a] -> [a]
invertir [] = []
invertir (x:xs) = invertir xs ++ [x]
```

```haskell
invertir[5,4,3,2,1]
```

```haskell
cuadrado :: Int -> Int
cuadrado x = x * x
 
suma :: Int -> Int -> Int
suma x y = x + y
 
sumaCuadrados :: Int -> Int -> Int
sumaCuadrados x y = suma (cuadrado x) (cuadrado y)
```

```haskell
sumaCuadrados 3 4
sumaCuadrados 1 1
sumaCuadrados 0 5
sumaCuadrados 2 2
```

```haskell
duplicar :: Int -> Int
duplicar x = x * 2
 
dobleLista :: [Int] -> [Int]
dobleLista xs = map duplicar xs
```

```haskell
dobleLista [1,2,3,4]  
```

```haskell
clasificarNota :: Int -> String
clasificarNota n
  | n < 3     = "Reprobado"
  | n <= 4    = "En recuperación"
  | otherwise = "Aprobado"
```

```haskell
clasificarNota 2   
clasificarNota 3   
clasificarNota 5   
```

```haskell
fibonacci :: Int -> Int
fibonacci 0 = 0
fibonacci 1 = 1
fibonacci n = fibonacci (n - 1) + fibonacci (n - 2)
```

```haskell
fibonacci 4   
fibonacci 6   
```

```haskell
doble :: Int -> Int
doble = \x -> x * 2
 
cubo :: Int -> Int
cubo = \x -> x * x * x
 
mitad :: Float -> Float
mitad = \x -> x / 2
 
esPositivo :: Int -> Bool
esPositivo = \x -> x > 0
 
enIntervalo :: Int -> Bool
enIntervalo = \x -> x >= 10 && x <= 20
```

```haskell
doble 5              
cubo 3               
mitad 7              
esPositivo (-4)      
enIntervalo 15       
enIntervalo 25      
```

```haskell
sumaLambda :: Int -> Int -> Int
sumaLambda = \x y -> x + y

multLambda :: Int -> Int -> Int
multLambda = \x y -> x * y
 
potenciaLambda :: Int -> Int -> Int
potenciaLambda = \x y -> x ^ y
 
maximoLambda :: Int -> Int -> Int
maximoLambda = \x y -> if x > y then x else y
  
igualesLambda :: Int -> Int -> Bool
igualesLambda = \x y -> x == y
 
```

```haskell
sumaLambda 3 4        
multLambda 3 4        
potenciaLambda 2 5    
maximoLambda 7 3      
igualesLambda 5 5     
igualesLambda 5 6     
```
